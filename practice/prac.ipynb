```python
import numpy as np
import pandas as pd
```

```python
#movieLens 1M dataset
```

```python
unames = ['user_id', 'gender', 'age', 'occupation', 'zip']
users = pd.read_table('../datasets/movielens/users.dat', sep='::', header=None, names=unames, engine='python')
```

```python
rnames = ['user_id', 'movie_id', 'rating', 'timestamp']
ratings = pd.read_table('../datasets/movielens/ratings.dat', sep='::', header=None, names=rnames, engine='python')
```

```python
mnames = ['movie_id', 'title', 'genres']
movies = pd.read_table('../datasets/movielens/movies.dat', sep='::', header=None, names=mnames, engine='python')
```

```python
users.head()
```

```python
data = pd.merge(pd.merge(ratings, users), movies)
```

```python
data.iloc[0]
```

```python
mean_ratings = data.pivot_table('rating', index = 'title', columns= 'gender', aggfunc= 'mean')
```

```python
mean_ratings
```

```python
ratings_title = data.groupby('title').size()
```

```python
ratings_title
```

```python
active_titles = ratings_title.index[ratings_title >= 250]
```

```python
active_titles
```

```python
mean_ratings = mean_ratings.loc[active_titles]
```

```python
mean_ratings
```

```python
top_female_ratings = mean_ratings.sort_values(by='F', ascending=False)
```

```python
top_female_ratings
```

```python
#measuring rating disagreement
mean_ratings['diff'] = mean_ratings['M'] - mean_ratings['F']
```

```python
sorted_by_diff = mean_ratings.sort_values(by='diff')
```

```python
sorted_by_diff
```

```python
sorted_by_diff[::-1]
```

```python
#find the movies that have the largest rating difference independent of gender
rating_std_by_title = data.groupby('title')['rating'].std()
```

```python
rating_std_by_title.loc[active_titles].sort_values(ascending=False) 
```
