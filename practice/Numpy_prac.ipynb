```python
## random work with 1_000 steps with 1 or -1 with equal prob
#pure python way
import random
position = 0
walk = [position]
nsteps = 1000
for _ in range(nsteps):
    step = 1 if random.randint(0,1) else -1
    position += step
    walk.append(position)
```

```python
walk
```

```python
plt.plot(walk)
```

```python
#multiple random walks
nwalks = 5000
nsteps = 1000
```

```python
draws = rng.integers(0 , 2, size=(nwalks, nsteps))
```

```python
steps = np.where(draws > 0, 1, -1)
```

```python
walks = steps.cumsum(axis = 1)
```

```python
walks
```

```python
walks.min()
```

```python
walks.max()
```

```python
hits30 = (np.abs(walks) >= 30).any(axis = 1)
```

```python
hits30
```

```python
hits30.sum()
```

```python
(np.abs(walks) >= 30).argmax(axis=1)
```

```python
crossing_times = (np.abs(walks[hits30])>=30).argmax(axis = 1)
```

```python
crossing_times.mean()
```
