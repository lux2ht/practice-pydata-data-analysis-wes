```python
rng = np.random.default_rng(seed=12345)
draws = rng.standard_normal(1000)
```

```python
draws[:5]
```

```python
bins = pd.qcut(draws, 4, labels= ['Q1', 'Q2', 'Q3', 'Q4'])
```

```python
bins
```

```python
bins.codes[:5]
```

```python
bins = pd.Series(bins, name= 'quartile')
```

```python
bins
```

```python
results = (
    pd.Series(draws)
    .groupby(bins, observed=False)
    .agg(['count', 'min', 'max'])
    .reset_index()
)
```

```python
results
```

```python
results['quartile']
```
