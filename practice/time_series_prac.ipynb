```python
# moving window functions
```

```python
close_px_all = pd.read_csv('../examples/stock_px.csv', index_col= 0, parse_dates= True)
```

```python
close_px = close_px_all[["AAPL", "MSFT", "XOM"]]
```

```python
close_px.isna().sum().sum()
```

```python
close_px = close_px.resample('B').ffill()
```

```python
close_px.describe()
```

```python
close_px['AAPL'].plot()
close_px['AAPL'].rolling(250).mean().plot()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure()
```

```python
std250 = close_px['AAPL'].pct_change().rolling(250, min_periods = 10).std()
```

```python
std250.plot()
```

```python
expanding_mean = std250.expanding().mean()
```

```python
expanding_mean.plot()
```

```python
plt.style.use('grayscale')
```

```python
close_px.rolling(60).mean().plot(logy = True)
```

```python
close_px.rolling("20D").mean().plot(logy = True)
```

```python
appl_px = close_px['AAPL']['2006':'2007']
```

```python
ma30 = appl_px.rolling(30, min_periods= 20).mean()
```

```python
ewma30 = appl_px.ewm(span= 30).mean()
```

```python
appl_px.plot(style = 'k-', label = 'Price')
ma30.plot(style = 'k--', label = 'Simple Moving Avg')
ewma30.plot(style = 'k-', label = 'EW MA')
plt.legend()
```

```python
spx_px = close_px_all['SPX']
```

```python
spx_rets = spx_px.pct_change()
```

```python
returns = close_px.pct_change()
```

```python
corr = returns['AAPL'].rolling(125, min_periods= 100). corr(spx_px)
```

```python
corr.plot()
```

```python
corr = returns.rolling(125, min_periods= 100). corr(spx_px)
```

```python
corr.plot()
```

```python
from scipy.stats import percentileofscore
```

```python
def score_at_2percent(x):
    return percentileofscore(x, 0.02)
```

```python
result = returns['AAPL'].rolling(250).apply(score_at_2percent)
```

```python
result.plot()
```
